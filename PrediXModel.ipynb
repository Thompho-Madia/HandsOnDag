```python
import yfinance as yf
```

## Data Extraction

```python
def get_forex_data(pair: str, period="1y", interval="1d"):
    ticker = f"{pair.upper()}=X"
    df = yf.download(ticker, period=period, interval=interval)
    if df.empty:
        print("⚠️ No data found. Check if the pair is valid.")
    return df

# Example usage
df = get_forex_data("EURUSD", period="6mo")
print(df.tail())
```
